# Aurora · Colab GPU worker

Turn a Colab GPU into an Aurora backend (lipsync by default — fits the free tier).

**Before running:**
1. **Runtime → Change runtime type** → Hardware accelerator = **GPU**.
2. **Secrets** (key icon, left sidebar): add `NGROK_AUTHTOKEN`, `NGROK_STATIC_DOMAIN`, `AURORA_URL`, `AURORA_REGISTER_SECRET` (optional: `AURORA_WORKER_TOKEN`, `AURORA_TASKS`) and toggle **notebook access ON** for each.
3. **Runtime → Run all**.

See `workers/colab/README.md` for where to get each secret. The worker auto-registers in **Admin → Workers** and serves lip-sync jobs for free.

> `motion` (MimicMotion) needs a ≥24 GB GPU (Colab Pro+ A100) — not the free tier. `lipsync` fits any free Colab GPU.

```python
# Aurora · Colab GPU worker — one-cell launcher.
# Runtime -> Change runtime type -> GPU. Secrets (key icon): NGROK_AUTHTOKEN,
# NGROK_STATIC_DOMAIN, AURORA_URL, AURORA_REGISTER_SECRET (+ optional
# AURORA_WORKER_TOKEN, AURORA_TASKS). Toggle notebook access ON for each, then
# Runtime -> Run all. This fetches the launcher directly from your Aurora app
# (no GitHub access or public repo needed).
import os, urllib.request

try:
    from google.colab import userdata
    try:
        _v = userdata.get("AURORA_URL")
        if _v:
            os.environ["AURORA_URL"] = _v.strip()
    except Exception:
        pass
except Exception:
    pass

_aurora_url = os.environ.get("AURORA_URL", "").strip().rstrip("/")
if not _aurora_url:
    raise SystemExit(
        "AURORA_URL secret is not set. Add it in Secrets (key icon) "
        "(your Aurora app URL, e.g. https://your-app.replit.app) and re-run."
    )
_dst = "/content/aurora_worker_colab.py"
_url = f"{_aurora_url}/api/public/workers/files/colab_bootstrap.py"
urllib.request.urlretrieve(_url, _dst)
exec(open(_dst).read())
```
